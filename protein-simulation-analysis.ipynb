# Post-simulation analysis for protein MD

This notebook is the companion to ``protein-simulation-setup.ipynb`` and assumes a production trajectory has already been generated.

It performs:
* energy sanity checks from the ``.edr`` file,
* trajectory alignment and PBC imaging,
* interactive geometric measurements (distances, angles, dihedrals),
* secondary-structure timeline (DSSP),
* Ramachandran analysis,
* and exports a clean, aligned trajectory/topology pair for visualisation in **Molstar**.

All plots are saved to the ``analysis_output/`` directory in publication-ready format.

## Configuration

Set the same ``name`` you used in the setup notebook, then derived file paths.
The setup notebook writes production coordinates for the ``Protein`` group only, so this notebook creates or reuses a matching protein-only structure file before loading the trajectory.
Adjust ``timestep_ps`` if your integration step differs from the standard 2 fs.

```python
name = "protein"

out_dir = "analysis_output"

traj_path = name + ".xtc"
full_structure_path = name + ".gro"
structure_path = f"{out_dir}/{name}-structure.gro"
edr_path = name + ".edr"
tpr_path = name + ".tpr"

timestep_ps = 2.0  # integration step in ps
```

## Imports

```python
from pathlib import Path

import gromacs as gmx
import mdanalysis_utils as mdu
import nglview as nv
import numpy as np
from IPython.display import Image, display
from matplotlib import pyplot as plt


def show_plot(path):
    display(Image(filename=str(path)))

Path(out_dir).mkdir(parents=True, exist_ok=True)
```

## Energy sanity checks

Parse the GROMACS ``.edr`` energy file and plot temperature, pressure, and potential energy versus time.
We rely on ``gmx energy`` to write xvg files, then load the data with NumPy.

```python
# Extract energy terms to xvg files.
gmx.energy(f=edr_path, o=str(Path(out_dir) / "temp.xvg"), input="Temperature")
gmx.energy(f=edr_path, o=str(Path(out_dir) / "press.xvg"), input="Pressure")
gmx.energy(f=edr_path, o=str(Path(out_dir) / "pot.xvg"), input="Potential")
```

```python
def _load_xvg(path: Path) -> None:
    """Load an xvg file, skipping Gracecomment lines."""
    return np.loadtxt(path, comments=["#", "@"])


temp = _load_xvg(Path(out_dir) / "temp.xvg")
press = _load_xvg(Path(out_dir) / "press.xvg")
pot = _load_xvg(Path(out_dir) / "pot.xvg")
temperature_plot = Path(out_dir) / "temperature.png"
pressure_plot = Path(out_dir) / "pressure.png"
potential_plot = Path(out_dir) / "potential.png"

mdu.plot_energy(
    temp[:, 0] / 1000.0,
    temp[:, 1],
    title="Temperature",
    ylabel="Temperature (K)",
    outpath=temperature_plot,
)
mdu.plot_energy(press[:, 0] / 1000.0, press[:, 1], title="Pressure", ylabel="Pressure (bar)", outpath=pressure_plot)
mdu.plot_energy(
    pot[:, 0] / 1000.0,
    pot[:, 1],
    title="Potential energy",
    ylabel="Potential energy (kJ/mol)",
    outpath=potential_plot,
)

for plot_path in (temperature_plot, pressure_plot, potential_plot):
    show_plot(plot_path)
```

## Trajectory preprocessing

Load the raw trajectory, **align** the protein to the first frame, and attempt
**PBC imaging** so that molecules are whole.  A quick ``nglview`` preview lets you
inspect the aligned trajectory inline.

```python
def _matching_existing_structure(candidates, trajectory_path):
    trajectory_atoms = mdu.xtc_atom_count(trajectory_path)
    for candidate in candidates:
        candidate = Path(candidate)
        if candidate.exists() and mdu.topology_atom_count(candidate) == trajectory_atoms:
            return candidate
    return None


def ensure_matching_structure(trajectory_path, structure_path, full_structure_path, tpr_path, selection="Protein"):
    structure_path = Path(structure_path)
    tpr_path = Path(tpr_path)
    structure_candidates = [Path(full_structure_path), Path("npt.gro"), Path("nvt.gro"), Path("em.gro")]

    existing = _matching_existing_structure([structure_path, Path(name + "-vis.gro"), *structure_candidates], trajectory_path)
    if existing is not None:
        return str(existing)

    if not tpr_path.exists():
        raise FileNotFoundError(f"Cannot create {structure_path}: production TPR file not found: {tpr_path}")

    source_structure = next((path for path in structure_candidates if path.exists()), None)
    if source_structure is None:
        raise FileNotFoundError(
            f"Cannot create {structure_path}: no coordinate structure found. Expected one of "
            f"{', '.join(str(path) for path in structure_candidates)}."
        )

    index_path = structure_path.with_suffix(".ndx")
    gmx.select(s=str(tpr_path), on=str(index_path), select=selection)
    gmx.trjconv(s=str(tpr_path), f=str(source_structure), n=str(index_path), o=str(structure_path), input=selection)
    return str(structure_path)


structure_path = ensure_matching_structure(traj_path, structure_path, full_structure_path, tpr_path)
traj_info = mdu.diagnose_trajectory_inputs(traj_path, structure_path)
print(
    "Trajectory/topology check: "
    f"{traj_info['trajectory_atoms']} trajectory atoms, "
    f"{traj_info['topology_atoms']} topology atoms"
)

traj = mdu.load_trajectory_checked(traj_path, structure_path)
print(f"Loaded {traj.n_frames} frames, {traj.n_atoms} atoms, {traj.n_residues} residues")
```

```python
traj_aligned = mdu.align_trajectory(traj)
traj_aligned = mdu.image_molecules(traj_aligned)
print("Alignment and imaging complete.")
```

```python
if distance_pairs:
    t, dists, labels = mdu.measure_distance(traj_aligned, distance_pairs, timestep_ps=timestep_ps)

    fig, ax = plt.subplots(figsize=(6, 3.5))
    for i, label in enumerate(labels):
        ax.plot(t, dists[:, i], label=label, linewidth=0.8)
    ax.set_xlabel("Time (ns)")
    ax.set_ylabel("Distance (nm)")
    ax.set_title("Pairwise distances")
    ax.legend(fontsize=8, frameon=False)
    ax.grid(True, linestyle="--", linewidth=0.4, alpha=0.7)
    fig.savefig(Path(out_dir) / "distances.png")
    plt.show()
else:
    print("No distance pairs configured; skipping distance plot.")
```

```python
if angle_triples:
    t, angles, labels = mdu.measure_angle(traj_aligned, angle_triples, timestep_ps=timestep_ps)

    fig, ax = plt.subplots(figsize=(6, 3.5))
    for i, label in enumerate(labels):
        ax.plot(t, angles[:, i], label=label, linewidth=0.8)
    ax.set_xlabel("Time (ns)")
    ax.set_ylabel("Angle (°)")
    ax.set_title("Angles")
    ax.legend(fontsize=8, frameon=False)
    ax.grid(True, linestyle="--", linewidth=0.4, alpha=0.7)
    fig.savefig(Path(out_dir) / "angles.png")
    plt.show()
else:
    print("No angle triples configured; skipping angle plot.")
```

```python
if dihedral_quads:
    t, dihs, labels = mdu.measure_dihedral(traj_aligned, dihedral_quads, timestep_ps=timestep_ps)

    fig, ax = plt.subplots(figsize=(6, 3.5))
    for i, label in enumerate(labels):
        ax.plot(t, dihs[:, i], label=label, linewidth=0.8)
    ax.set_xlabel("Time (ns)")
    ax.set_ylabel("Dihedral (°)")
    ax.set_title("Dihedrals")
    ax.legend(fontsize=8, frameon=False)
    ax.grid(True, linestyle="--", linewidth=0.4, alpha=0.7)
    fig.savefig(Path(out_dir) / "dihedrals.png")
    plt.show()
else:
    print("No dihedral quads configured; skipping dihedral plot.")
```

## Structure quality — DSSP timeline

Run DSSP on every frame and display a heat-map of secondary-structure evolution.
Helices, sheets, and coil/other are shown on a colour scale.

```python
dssp_plot = Path(out_dir) / "dssp_timeline.png"
time_ns, residues, ss_codes = mdu.dssp_timeline(traj_aligned, outpath=dssp_plot)
print(f"DSSP computed for {traj_aligned.n_frames} frames x {len(residues)} residues")
show_plot(dssp_plot)
```

Extract backbone φ / ψ dihedrals for every frame and draw a 2-D histogram.

```python
ramachandran_plot = Path(out_dir) / "ramachandran.png"
phi, psi = mdu.ramachandran(
    traj_aligned,
    residue_indices=None,  # set to e.g. [10, 11, 12] to zoom in
    outpath=ramachandran_plot,
)
print(f"Collected {len(phi)} (phi, psi) measurements")
show_plot(ramachandran_plot)
```
